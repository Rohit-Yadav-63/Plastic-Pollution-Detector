```python
!pip install roboflow

from roboflow import Roboflow
rf = Roboflow(api_key="YOUR_API_KEY")
project = rf.workspace("edwin-daza-saavedra-s-workspace").project("plastic-waste-ag4eg")
version = project.version(2)
dataset = version.download("yolov8")
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
model.train(data=f"{dataset.location}/data.yaml", epochs=15, imgsz=512, fraction=0.5)

import glob, shutil
from google.colab import files
files.download("/content/runs/detect/train/weights/best.pt")

imgs = sorted(glob.glob("/content/*/valid/images/*.jpg"))[:8]
model.predict(source=imgs, save=True, conf=0.25)

shutil.make_archive("results_pack", "zip", "/content/runs/detect")
files.download("results_pack.zip")
```
